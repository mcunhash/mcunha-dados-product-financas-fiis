```sql
%sql
CREATE TABLE IF NOT EXISTS product_dev.financas.analytics_performance_fiis (
  ticker STRING NOT NULL,
  categoria_fii STRING NOT NULL,
  nome_fundo STRING NOT NULL,
  preco_atual DOUBLE NOT NULL,
  p_vp DOUBLE NOT NULL,
  dividend_yield_12m DOUBLE NOT NULL,
  score_final DOUBLE NOT NULL,
  posicao_ranking INT NOT NULL,
  data_selecao DATE NOT NULL,
  data_carga TIMESTAMP NOT NULL
)
USING DELTA;
```

```sql
%sql
COMMENT ON TABLE product_dev.financas.analytics_performance_fiis 
IS 'Tabela da camada Gold contendo a seleção consolidada do desempenho dos melhores FIIs de cada categoria (Tijolo, Papel e FoF) de acordo com o modelo de scoring analítico.';
```
